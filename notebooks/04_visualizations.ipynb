## Visualizations

```python
import pandas as pd
import altair as alt
```

```python
hms = pd.read_csv("../data/processed/hms_clean.csv")
```

```python
hms["survey_year"].value_counts()
```

```python
flourish_by_year = hms[["flourish", "survey_year"]].groupby("survey_year")["flourish"].mean().reset_index()
```

```python
flourish_by_year
```

```python
alt.Chart(flourish_by_year).mark_bar().encode(
    alt.X("survey_year:O", title="Survey Year"),
    alt.Y("flourish:Q", title="Flourishing and Positive Well-Being"),
    tooltip=["flourish"]
)
```

```python
deprawsc_by_year = hms[["deprawsc", "survey_year"]].groupby("survey_year")["deprawsc"].mean().reset_index()
```

```python
alt.Chart(deprawsc_by_year).mark_bar().encode(
    alt.X("survey_year:O", title="Survey Year"),
    alt.Y("deprawsc:Q", title="Depression Symptom Score"),
    tooltip=["deprawsc"]
)
```

```python
anx_score_by_year = hms[["anx_score", "survey_year"]].groupby("survey_year")["anx_score"].mean().reset_index()
```

```python
alt.Chart(anx_score_by_year).mark_bar().encode(
    alt.X("survey_year:O", title="Survey Year"),
    alt.Y("anx_score:Q", title="Anxiety Symptom Score"),
    tooltip=["anx_score"]
)
```
