## Visualizations

```python
# imports
import pandas as pd
import altair as alt
```

```python
# load data
hms = pd.read_csv("../data/processed/hms_clean.csv")
```

### Outcomes Across Years

```python
# flourish by year
flourish_by_year = hms[["flourish", "survey_year"]].groupby("survey_year")["flourish"].mean().reset_index()

# line plot for flourish
flourish_by_year_plot = alt.Chart(flourish_by_year).mark_line().encode(
    alt.X("survey_year:O", title="Survey Year"),
    alt.Y("flourish:Q", title="Flourishing and Positive Well-Being Score", scale=alt.Scale(domain=[40, 50])),
    tooltip=["flourish"]
).properties(width=250, title="Average Flourish and Positive Well-Being Score per Year")
```

```python
# depression score by year
deprawsc_by_year = hms[["deprawsc", "survey_year"]].groupby("survey_year")["deprawsc"].mean().reset_index()

# line plot for deprawsc
deprawsc_by_year_plot = alt.Chart(deprawsc_by_year).mark_line().encode(
    alt.X("survey_year:O", title="Survey Year"),
    alt.Y("deprawsc:Q", title="Depression Symptom Score", scale=alt.Scale(domain=[7, 10])),
    tooltip=["deprawsc"]
).properties(width=250, title="Average Depression Sympotom Score per Year")
```

```python
# anxiety score by year
anx_score_by_year = hms[["anx_score", "survey_year"]].groupby("survey_year")["anx_score"].mean().reset_index()

# line plot for anx_score
anx_score_by_year_plot = alt.Chart(anx_score_by_year).mark_line().encode(
    alt.X("survey_year:O", title="Survey Year"),
    alt.Y("anx_score:Q", title="Anxiety Symptom Score", scale=alt.Scale(domain=[6.5, 9])),
    tooltip=["anx_score"]
).properties(width=250, title="Average Anxiety Sympotom Score per Year")
```

```python
# plots next to eachother
flourish_by_year_plot | deprawsc_by_year_plot | anx_score_by_year_plot
```
