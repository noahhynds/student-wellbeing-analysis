## Database Creation

This notebook will be used to create the database that will allow for SQL based analysis

```python
import pandas as pd
import sqlite3
```

```python
# read data
hms_clean = pd.read_csv("../data/processed/hms_clean.csv")
```

```python
# create sqlite3 connection
connect = sqlite3.connect("../data/processed/hms_clean.db")
```

```python
# turn file to sql db

hms_clean.to_sql(name = "students", con = connect, if_exists = "replace", index = False)
```
