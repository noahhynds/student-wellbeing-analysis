## Comparisons Between Data

```python
# Difference between rows and columns
hms_summary = pd.DataFrame(
    {"Year": ["2022-2023", "2023-2024", "2024-2025", "2025-2026"],
    "Rows": [hms_23.shape[0], hms_24.shape[0], hms_25.shape[0], hms_26.shape[0]],
    "Columns": [hms_23.shape[1], hms_24.shape[1], hms_25.shape[1], hms_26.shape[1]]}
)
hms_summary
```

The amount of observations and variables varies between the years with 2023-2024 having the most participants and 2025-2026 having the least amount of participants. The variation of columns means that the structure of the survey has changed from year-to-year. Before combining, cleaning, and analyzing the datasets further, the variables will need to be compared across each year.

```python
# Compare columns across years

# Columns
hms_23_cols = hms_23.columns.to_list()
hms_24_cols = hms_24.columns.to_list()
hms_25_cols = hms_25.columns.to_list()
hms_26_cols = hms_26.columns.to_list()

# Compare using set intersection
common_cols = list(set(hms_23_cols) & set(hms_24_cols) & set(hms_25_cols) & set(hms_26_cols))
len(common_cols)
```

The datasets share 1067 variables across all four years the survey was done. These variables will act as a starting point for identifying factors and outcome that will be used in the combined analysis of the four years. Although these variables share the same name, they must have the same meaning, essentially the same survey question, and same response coding across all four years. This analysis will be done using the data dictionaries from the HMS codebooks from 2022-2023, 2023-2024, 2024-2025, and 2025-2026. Variables will be split into: Outcome, Main factor, control, identifier, and survery weight. Outcome variables are ones we are trying to explain, main factors are going to be used to see if it is associated with the outcome, control variables are background charasteristics that can be accounetd for when analyzing relationships between main factors and outcomes, identifiers will be used to know where the observation came from, and survey weights will be kept for use later in statistical analysis to account for survey design and produce weighted estimates.

## Data Dictionary

| Variable | Category | Role | Description | Coding / Range |
|---|---|---|---|---|
| `flourish` | Mental well-being | Outcome | Flourishing Scale total score measuring positive mental well-being | 8–56; higher values indicate greater flourishing |
| `deprawsc` | Depression | Outcome | PHQ-9 total depression symptom score | 0–27; higher values indicate greater depressive symptom severity |
| `anx_score` | Anxiety | Outcome | GAD-7 total anxiety symptom score | 0–21; higher values indicate greater anxiety symptom severity |
| `fincur` | Financial | Main factor | Self-reported current financial stress | 1 = Always stressful, 2 = Often stressful, 3 = Sometimes stressful, 4 = Rarely stressful, 5 = Never stressful |
| `finpast` | Financial | Main factor | Self-reported financial stress while growing up | 1 = Always stressful, 2 = Often stressful, 3 = Sometimes stressful, 4 = Rarely stressful, 5 = Never stressful |
| `belong1` | Social / Campus | Main factor | Agreement with the statement "I see myself as a part of the campus community" | 1 = Strongly agree through 6 = Strongly disagree |
| `discrim_race` | Campus experience | Main factor | Experienced unfair treatment at school because of race/ethnicity in the past 12 months | 0 = No, 1 = Yes |
| `discrim_culture` | Campus experience | Main factor | Experienced unfair treatment at school because of cultural background in the past 12 months | 0 = No, 1 = Yes |
| `discrim_gender` | Campus experience | Main factor | Experienced unfair treatment at school because of gender in the past 12 months | 0 = No, 1 = Yes |
| `discrim_sexual` | Campus experience | Main factor | Experienced unfair treatment at school because of sexual orientation in the past 12 months | 0 = No, 1 = Yes |
| `alc_any`* | Behavioural | Main factor | Any alcohol consumption during the past 2 weeks | 0 = No, 1 = Yes |
| `age` | Demographic | Control | Respondent age | Age in years |
| `international` | Demographic | Control | Whether the respondent is an international student | 0 = No, 1 = Yes |
| `educ_par1` | Socioeconomic | Control | Highest education level completed by first parent, step-parent, or guardian | 1 = 8th grade or lower; 2 = 9th–12th grade, no high school degree; 3 = High school degree; 4 = Some college, no college degree; 5 = Associate's degree; 6 = Bachelor's degree; 7 = Graduate degree; 8 = Don't know |
| `educ_par2` | Socioeconomic | Control | Highest education level completed by second parent, step-parent, or guardian | 1 = 8th grade or lower; 2 = 9th–12th grade, no high school degree; 3 = High school degree; 4 = Some college, no college degree; 5 = Associate's degree; 6 = Bachelor's degree; 7 = Graduate degree; 8 = Don't know |
| `enroll` | Academic | Control | Respondent's enrollment status | 1 = Full-time student, 2 = Part-time student, 3 = Other |
| `survey_year` | Dataset | Identifier / Control | Academic year in which the respondent completed the HMS survey | 2022–2023, 2023–2024, 2024–2025, or 2025–2026 |
| `nrweight` | Survey design | Survey weight | HMS non-response weight used when producing weighted estimates | Continuous survey weight; not used as a predictor |

\*`alc_any` uses consistent wording and coding across all four survey years but whether students were asked this question depended on if their school used the Substance Use module.
